# Galectin-8 Specificity Analysis

This notebook characterises the glycan-binding specificity of **Galectin-8** (with focus on its C-terminal carbohydrate-recognition domain, C-CRD) and compares it directly against Galectin-3 — the lectin chosen as the anchor domain in the glEYEco fusion protein.

The central question answered here is:

> *What is the interaction between Galectin-8 and the core-1 T-antigen (Gal(β1-3)GalNAc), and how does this compare with Galectin-3's affinity for the same epitope and for ocular surface glycans generally?*

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns

# ── Core glycowork datasets ───────────────────────────────────────────────
from glycowork.glycan_data.loader import df_glycan, glycan_binding

# ── Motif annotation ──────────────────────────────────────────────────────
from glycowork.motif.annotate import annotate_dataset

# ── Visualisation ─────────────────────────────────────────────────────────
from glycowork import GlycoDraw

print(f"df_glycan      : {df_glycan.shape[0]:,} glycan entries,  {df_glycan.shape[1]} columns")
print(f"glycan_binding : {glycan_binding.shape[0]:,} proteins × {glycan_binding.shape[1]-2:,} glycan columns")
```

### 1.1 Shared helpers & constants

All filtering functions and keyword lists are defined once here so every downstream section can call them without redefinition.


```python
# ── Helper: flexible column-value matching ────────────────────────────────
def col_contains_any(col_value, terms):
    """
    Return True if any element of col_value (str or list) contains
    any of the given terms (case-insensitive substring match).
    """
    if col_value is None or isinstance(col_value, float):
        return False
    items = col_value if isinstance(col_value, list) else [str(col_value)]
    return any(term.lower() in item.lower() for item in items for term in terms)


# ── Ocular tissue keywords (mirrors the main notebook) ───────────────────
OCULAR_TISSUE_TERMS = [
    'eye', 'cornea', 'tears', 'ocular',
    'lacrimal', 'conjunctiva', 'lens'
]

# ── Glycan motif patterns (IUPAC-condensed regex) ─────────────────────────
T_ANTIGEN_PATTERN = r'Gal\(b1-3\)GalNAc'   # Core-1 / T-antigen
LACNAC_PATTERN    = r'Gal\(b1-4\)GlcNAc'   # Type-2 LacNAc
BETA_GAL_PATTERN  = r'Gal\(b'               # Any β-galactoside

# ── Protein names as stored in glycan_binding ────────────────────────────
GAL3_NAME  = 'Galectin-3'
GAL8C_NAME = 'Galectin-8 C-terminal CRD'

# ── Pre-build ocular tissue subset (used in Sections 5 & 7) ──────────────
tissue_mask      = df_glycan['tissue_sample'].apply(
    lambda x: col_contains_any(x, OCULAR_TISSUE_TERMS)
)
ocular_tissue_df = df_glycan[tissue_mask].copy()
ocular_glycan_set = set(ocular_tissue_df['glycan'].tolist())

# ── Glycan column list (all columns that are not protein metadata) ─────────
glycan_cols = [c for c in glycan_binding.columns if c not in ('protein', 'target')]

print(f"Ocular tissue glycans loaded : {len(ocular_tissue_df)}")
print(f"Glycan columns in binding db : {len(glycan_cols):,}")
```

## Section 2 — Galectin-8 entries in `glycan_binding`

The `glycan_binding` dataset is a protein × glycan matrix of log-transformed binding scores derived from glycan microarray experiments (Consortium for Functional Glycomics, CFG).  
Each row is one protein; each column is a glycan sequence. Values are log-scale relative fluorescence units (RFU); NaN = not measured on that array.

Galectin-8 is a tandem-repeat galectin with **two structurally distinct CRDs**:
- **N-terminal CRD** — preferentially binds 3′-sialylated and sulfated glycans  
- **C-terminal CRD** — preferentially binds poly-LacNAc and blood-group epitopes (the β-galactoside–recognising domain most relevant here)

We retrieve every row whose protein name contains "galectin-8" to capture both domains if present.


```python
# ── Retrieve all Galectin-8 entries ──────────────────────────────────────
gal8_df = glycan_binding[
    glycan_binding['protein'].str.contains('galectin-8', case=False, na=False)
].copy()

print("Galectin-8 entries found in glycan_binding:")
for name in gal8_df['protein'].tolist():
    print(f"  • {name}")

print(f"\nTotal entries: {len(gal8_df)}")
```

### 2.1 Build per-isoform score series

For each Galectin-8 entry we transpose its row into a ranked `(glycan → score)` Series, dropping glycans that were not measured (NaN). This format is used in all downstream sections.


```python
# ── Build score series for each Galectin-8 isoform ──────────────────────
gal8_profiles = {}

for _, row in gal8_df.iterrows():
    name   = row['protein']
    scores = (
        pd.Series(row[glycan_cols], name='binding_score')
        .dropna()
        .sort_values(ascending=False)
    )
    gal8_profiles[name] = scores

    pos = (scores > 0).sum()
    neg = (scores <= 0).sum()
    print(f"{'─'*55}")
    print(f"  {name}")
    print(f"  Glycans measured : {len(scores)}")
    print(f"  Positive binders : {pos}   (score > 0)")
    print(f"  Background/neg.  : {neg}")
    print(f"\n  Top binders:")
    print(scores.head(5).to_string())

# Convenience alias for the C-terminal CRD used throughout
if GAL8C_NAME not in gal8_profiles:
    raise KeyError(
        f"'{GAL8C_NAME}' not found. "
        f"Available: {list(gal8_profiles.keys())}"
    )
gal8c_scores = gal8_profiles[GAL8C_NAME]
```

### 2.2 Full binding profile — Galectin-8 C-terminal CRD

Plot all measured glycans for the C-terminal CRD, coloured green (positive binding) or red (background / negative).


```python
fig, ax = plt.subplots(figsize=(9, max(4, len(gal8c_scores) * 0.38)))

colors = ['#2ecc71' if s > 0 else '#e74c3c' for s in gal8c_scores]
gal8c_scores.plot(kind='barh', ax=ax, color=colors, edgecolor='white')

ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Log-scale binding score (RFU)')
ax.set_title(f'{GAL8C_NAME}  —  full glycan binding profile')
ax.invert_yaxis()

pos_p = mpatches.Patch(color='#2ecc71', label='Positive binding (score > 0)')
neg_p = mpatches.Patch(color='#e74c3c', label='Background / negative')
ax.legend(handles=[pos_p, neg_p], loc='lower right')

plt.tight_layout()
plt.savefig('fig_gal8c_binding_profile.png', dpi=150)
plt.show()
print("Figure saved: fig_gal8c_binding_profile.png")
```

## Section 3 — Core-1 T-antigen interaction

The **core-1 T-antigen** (`Gal(β1-3)GalNAc`) is a key O-glycan epitope on corneal mucins (MUC5AC, MUC16) and is a well-established Galectin-3 ligand. Here we query Galectin-8 C-CRD affinity for this structure using three complementary approaches:

| Layer | Method | What it tells us |
|-------|--------|-----------------|
| 1 | Direct column look-up in `glycan_binding` | Whether the CFG array included T-antigen–containing glycans that were also measured against Gal8c |
| 2 | Motif search within Gal8c-measured glycans | Whether any glycans *already measured* for Gal8c happen to carry T-antigen |
| 3 | T-antigen in ocular `df_glycan` | How many ocular tissue glycans carry this epitope (clinical relevance) |

A null result from Layer 1 indicates **array-design coverage**, not absence of binding — the CFG array does not include all possible glycan structures.


```python
# ── Layer 1: Direct column match in glycan_binding ───────────────────────
t_antigen_cols = [
    c for c in glycan_cols
    if pd.Series([c]).str.contains(T_ANTIGEN_PATTERN, regex=True).any()
]

print(f"Glycan columns in glycan_binding containing T-antigen motif: {len(t_antigen_cols)}")
if t_antigen_cols:
    for c in t_antigen_cols:
        print(f"  • {c}")
else:
    print("  None — the CFG array did not include undecorated T-antigen entries")
    print("  for this protein. See Layers 2 & 3 for the motif-level analysis.")
```

```python
# ── Layer 1 (continued): Galectin family scores for T-antigen columns ─────
if t_antigen_cols:
    galectin_all = glycan_binding[
        glycan_binding['protein'].str.contains('galectin', case=False, na=False)
    ].set_index('protein')

    t_ag_family = galectin_all[
        [c for c in t_antigen_cols if c in galectin_all.columns]
    ].copy()

    print("Galectin family binding scores for T-antigen–containing glycans:")
    print(t_ag_family.to_string())

    # Isolate Galectin-8 C-CRD specifically
    if GAL8C_NAME in t_ag_family.index:
        g8_t = t_ag_family.loc[GAL8C_NAME].dropna()
        print(f"\n{GAL8C_NAME} × T-antigen glycans ({len(g8_t)} measured):")
        for gly, score in g8_t.sort_values(ascending=False).items():
            print(f"  {score:+.4f}  {gly}")
    else:
        print(f"\n{GAL8C_NAME} not found in T-antigen columns — see Layer 2.")
else:
    print("Skipping Layer 1 extended analysis (no T-antigen columns found).")
```

```python
# ── Layer 2: T-antigen motif within Gal8c-measured glycans ───────────────
t_in_gal8c = gal8c_scores[
    gal8c_scores.index.str.contains(T_ANTIGEN_PATTERN, regex=True)
]

print(f"Glycans measured for Gal8c C-CRD that contain T-antigen motif: {len(t_in_gal8c)}")
if len(t_in_gal8c) > 0:
    print("\nBinding scores:")
    print(t_in_gal8c.to_string())
else:
    print("\nNone — the glycans measured for Gal8c C-CRD on this array do not include")
    print("an undecorated or simply decorated Gal(b1-3)GalNAc structure.")
    print("This is an array-coverage limitation, not a biological absence.")
```

```python
# ── Layer 3: T-antigen in ocular tissue (df_glycan) ──────────────────────
t_antigen_ocular = ocular_tissue_df[
    ocular_tissue_df['glycan'].str.contains(T_ANTIGEN_PATTERN, regex=True, na=False)
].reset_index(drop=True)

print(f"Ocular tissue glycans in df_glycan containing T-antigen: {len(t_antigen_ocular)}")
print("\nSequences:")
for i, g in enumerate(t_antigen_ocular['glycan'].tolist(), 1):
    print(f"  {i:>2}. {g}")

# Tissue label breakdown for these glycans
print("\nTissue associations:")
for _, row in t_antigen_ocular.iterrows():
    ts = row['tissue_sample']
    ts_str = ', '.join(ts) if isinstance(ts, list) else str(ts)
    print(f"  {row['glycan'][:60]:<60}  [{ts_str}]")
```

### 3.1 Motif annotation for ocular T-antigen glycans

Run `annotate_dataset` on the ocular T-antigen glycans to see what other structural features co-occur with this epitope at the ocular surface. This contextualises the binding environment any lectin-based anchor must navigate.


```python
if len(t_antigen_ocular) > 0:
    t_motif_df   = annotate_dataset(
        t_antigen_ocular['glycan'].tolist(),
        feature_set=['known', 'terminal'],
        condense=True
    )
    t_motif_freq = t_motif_df.sum().sort_values(ascending=False).head(15)

    print("Top 15 co-occurring motifs in ocular T-antigen glycans:")
    print(t_motif_freq.to_string())

    fig, ax = plt.subplots(figsize=(8, 5))
    t_motif_freq.plot(kind='barh', ax=ax, color='teal', edgecolor='white')
    ax.set_xlabel('Occurrence count')
    ax.set_title('Structural motifs co-occurring with T-antigen in ocular glycans')
    ax.invert_yaxis()
    plt.tight_layout()
    plt.savefig('fig_t_antigen_ocular_motifs.png', dpi=150)
    plt.show()
    print("Figure saved: fig_t_antigen_ocular_motifs.png")
else:
    print("No ocular T-antigen glycans found — motif annotation skipped.")
```

Draw the canonical core-1 T-antigen structure with `GlycoDraw` so the team can confirm the structural context being queried.


```python
# Sialylated T-antigen (core-1 O-glycan as found on corneal mucins)
T_ANTIGEN_EXAMPLE    = 'Neu5Ac(a2-3)Gal(b1-3)GalNAc'
T_ANTIGEN_BARE       = 'Gal(b1-3)GalNAc'

print("Core-1 T-antigen (bare):")
GlycoDraw(T_ANTIGEN_BARE)

print("\nSialylated core-1 T-antigen (as found on MUC5AC / MUC16):")
GlycoDraw(T_ANTIGEN_EXAMPLE, highlight_motif=T_ANTIGEN_BARE)
```

## Section 4 — Galectin-3 vs Galectin-8 C-CRD: head-to-head comparison

We pull Galectin-3 scores from `glycan_binding` using the same per-protein score-series approach, then merge the two series on their shared glycan index to create a directly comparable data frame.

**Why this matters for glEYEco:** If Galectin-3 and Galectin-8 C-CRD had near-identical specificity profiles, either could serve as the anchor domain. Differences in profile confirm that the domain choice is non-trivial and that Galectin-3 has properties specifically suited to the ocular target environment.


```python
# ── Retrieve Galectin-3 score series ─────────────────────────────────────
gal3_row    = glycan_binding[glycan_binding['protein'] == GAL3_NAME]

if gal3_row.empty:
    raise KeyError(f"'{GAL3_NAME}' not found in glycan_binding.")

gal3_scores = (
    gal3_row[glycan_cols]
    .T
    .dropna()
    .rename(columns={gal3_row.index[0]: GAL3_NAME})
)[GAL3_NAME]

print(f"Galectin-3 glycans measured : {len(gal3_scores)}")
print(f"  Positive binders          : {(gal3_scores > 0).sum()}")
print(f"\nGalectin-3 full profile:")
print(gal3_scores.to_string())
```

```python
# ── Merge on shared glycan index ─────────────────────────────────────────
comparison_df = pd.concat(
    [gal3_scores.rename(GAL3_NAME),
     gal8c_scores.rename(GAL8C_NAME)],
    axis=1
)

shared_df = comparison_df.dropna()   # glycans measured for BOTH
union_df  = comparison_df.copy()     # all glycans measured for either

print(f"Glycans measured for BOTH lectins : {len(shared_df)}")
print(f"Glycans measured for EITHER lectin: {len(union_df)}")

print("\nShared glycan comparison:")
if len(shared_df) > 0:
    print(shared_df.to_string())
else:
    print("  No glycans were measured for both lectins on the same array.")
    print("  This reflects array-design differences, not absence of binding.")

print("\nFull union (NaN = not measured for that lectin):")
print(union_df.to_string())
```

```python
# ── Side-by-side grouped bar chart ───────────────────────────────────────
if len(union_df) > 0:
    fig, ax = plt.subplots(
        figsize=(10, max(4, len(union_df) * 0.45))
    )

    x = range(len(union_df))
    w = 0.38

    g3_vals  = union_df[GAL3_NAME].fillna(0)
    g8c_vals = union_df[GAL8C_NAME].fillna(0)

    ax.barh(
        [i + w/2 for i in x], g3_vals,
        height=w, label='Galectin-3',
        color='#3498db', edgecolor='white'
    )
    ax.barh(
        [i - w/2 for i in x], g8c_vals,
        height=w, label='Galectin-8 C-terminal CRD',
        color='#e67e22', edgecolor='white'
    )

    ax.set_yticks(list(x))
    ax.set_yticklabels(union_df.index, fontsize=8)
    ax.axvline(0, color='black', linewidth=0.8)
    ax.set_xlabel('Log-scale binding score (RFU)')
    ax.set_title('Galectin-3 vs Galectin-8 C-terminal CRD  —  binding comparison')
    ax.legend()
    ax.invert_yaxis()

    # Annotate NaN bars so the reader knows they are absent, not zero
    for i, (g3, g8c) in enumerate(zip(union_df[GAL3_NAME], union_df[GAL8C_NAME])):
        if pd.isna(g3):
            ax.text(0.02, i + w/2, 'not measured', va='center', fontsize=7,
                    color='gray')
        if pd.isna(g8c):
            ax.text(0.02, i - w/2, 'not measured', va='center', fontsize=7,
                    color='gray')

    plt.tight_layout()
    plt.savefig('fig_gal3_vs_gal8c_comparison.png', dpi=150)
    plt.show()
    print("Figure saved: fig_gal3_vs_gal8c_comparison.png")
```

## Section 5 — Galectin-8 C-CRD × ocular tissue glycans

Cross-reference the Galectin-8 C-CRD measured glycan set with the ocular tissue subset of `df_glycan`. We check in three tiers:

1. **Exact match** — IUPAC string present in both datasets  
2. **LacNAc motif** — `Gal(β1-4)GlcNAc` within Gal8c-measured glycans  
3. **T-antigen motif** — `Gal(β1-3)GalNAc` within Gal8c-measured glycans  

This mirrors the approach in Section 3.4 of the main notebook (used for Galectin-3), allowing a like-for-like comparison.


```python
# ── Tier 1: Exact IUPAC matches ──────────────────────────────────────────
gal8c_ocular_exact = gal8c_scores[
    gal8c_scores.index.isin(ocular_glycan_set)
]
print(f"Exact overlap (Gal8c measured × ocular tissue): {len(gal8c_ocular_exact)}")
if len(gal8c_ocular_exact) > 0:
    print(gal8c_ocular_exact.to_string())
else:
    print("  None — IUPAC annotation differences between CFG array and df_glycan.")
```

```python
# ── Tier 2 & 3: Motif-level overlap in Gal8c-measured glycans ────────────
lacnac_in_gal8c    = gal8c_scores[
    gal8c_scores.index.str.contains(LACNAC_PATTERN, regex=True)
]
t_antigen_in_gal8c = gal8c_scores[
    gal8c_scores.index.str.contains(T_ANTIGEN_PATTERN, regex=True)
]

# Corresponding ocular counts (from df_glycan)
lacnac_ocular_count    = ocular_tissue_df['glycan'].str.contains(
    LACNAC_PATTERN, regex=True, na=False).sum()
t_antigen_ocular_count = ocular_tissue_df['glycan'].str.contains(
    T_ANTIGEN_PATTERN, regex=True, na=False).sum()

print("── Motif overlap summary ──────────────────────────────────────────────")
print(f"LacNAc [Gal(b1-4)GlcNAc]")
print(f"  Gal8c-measured glycans containing LacNAc   : {len(lacnac_in_gal8c)}")
print(f"  Ocular tissue glycans containing LacNAc    : {lacnac_ocular_count}")
print()
print(f"T-antigen [Gal(b1-3)GalNAc]")
print(f"  Gal8c-measured glycans containing T-antigen: {len(t_antigen_in_gal8c)}")
print(f"  Ocular tissue glycans containing T-antigen : {t_antigen_ocular_count}")
```

```python
# ── Show binding scores for LacNAc-containing glycans measured by Gal8c ──
if len(lacnac_in_gal8c) > 0:
    print("Galectin-8 C-CRD binding scores for LacNAc-containing glycans:")
    print(lacnac_in_gal8c.to_string())

    fig, ax = plt.subplots(figsize=(9, max(3, len(lacnac_in_gal8c) * 0.45)))
    colors = ['#2ecc71' if s > 0 else '#e74c3c' for s in lacnac_in_gal8c]
    lacnac_in_gal8c.plot(kind='barh', ax=ax, color=colors, edgecolor='white')
    ax.axvline(0, color='black', linewidth=0.8)
    ax.set_xlabel('Log-scale binding score (RFU)')
    ax.set_title('Galectin-8 C-CRD  ×  LacNAc-containing glycans')
    ax.invert_yaxis()
    plt.tight_layout()
    plt.savefig('fig_gal8c_lacnac_ocular.png', dpi=150)
    plt.show()
    print("Figure saved: fig_gal8c_lacnac_ocular.png")
else:
    print("No LacNAc-containing glycans in the Gal8c measured set.")
```

## Section 6 — Full galectin family heatmap

Visualise all four galectin family members present in `glycan_binding` (`Galectin-3`, `Galectin-7`, `Galectin-8 C-terminal CRD`, `Agalectin`) across every glycan for which at least one of them has a measured score.

This is the clearest way to show the team:
- Which glycan classes each galectin prefers
- How much Galectin-3 and Galectin-8 C-CRD overlap in specificity
- Whether Galectin-7 or Agalectin offer any advantage for ocular targeting


```python
# ── Retrieve all galectin family rows ─────────────────────────────────────
galectin_all_df = glycan_binding[
    glycan_binding['protein'].str.contains('galectin', case=False, na=False)
].set_index('protein')[glycan_cols]

# Keep only glycans measured for at least one galectin
galectin_measured = galectin_all_df.dropna(how='all', axis=1)

print(f"Galectin family members : {list(galectin_measured.index)}")
print(f"Glycans with ≥1 measurement: {galectin_measured.shape[1]}")
print("\nFull binding matrix:")
print(galectin_measured.to_string())
```

```python
if galectin_measured.shape[1] > 0:
    n_glycans = galectin_measured.shape[1]
    n_lectins  = galectin_measured.shape[0]

    # ── Hard cap: matplotlib cannot render images wider than 65535px.
    MAX_FIG_W = 40
    MAX_FIG_H = 20

    # ── Choose layout based on column count ──
    if n_glycans <= 30:
        fig_w    = min(MAX_FIG_W, max(12, n_glycans * 1.1))
        fig_h    = min(MAX_FIG_H, max(4,  n_lectins * 0.9))
        xticklabels = True
        annot       = True
        fontsize_x  = max(6, 11 - n_glycans // 5)   

    elif n_glycans <= 100:
        fig_w    = min(MAX_FIG_W, max(16, n_glycans * 0.28))
        fig_h    = min(MAX_FIG_H, max(4,  n_lectins * 0.9))
        xticklabels = True
        annot       = False
        fontsize_x  = 6

    else:
        # Large (e.g. full glycan_binding width): labels illegible anyway —
        # suppress them and show colour pattern only.
        fig_w    = min(MAX_FIG_W, 24)
        fig_h    = min(MAX_FIG_H, max(4, n_lectins * 0.9))
        xticklabels = False
        annot       = False
        fontsize_x  = 6

    fig, ax = plt.subplots(figsize=(fig_w, fig_h))

    sns.heatmap(
        galectin_measured.astype(float),
        ax=ax,
        cmap='RdYlGn',
        center=0,
        annot=annot,
        fmt='.2f' if annot else '',
        linewidths=0.3 if n_glycans <= 30 else 0,
        linecolor='white',
        cbar_kws={'label': 'Log-scale binding score (RFU)', 'shrink': 0.6},
        xticklabels=xticklabels,
        yticklabels=True,
    )

    subtitle = (
        '(green = positive binding, red = background/negative, white = not measured)'
        if n_glycans <= 100 else
        f'({n_glycans} glycans measured — x-axis labels suppressed for readability)'
    )
    ax.set_title(
        f'Galectin family  ×  glycan binding heatmap\n{subtitle}',
        pad=14, fontsize=11
    )
    ax.set_xlabel('Glycan (IUPAC-condensed)', labelpad=10)
    ax.set_ylabel('')
    plt.xticks(rotation=45, ha='right', fontsize=fontsize_x)
    plt.yticks(fontsize=10)
    plt.tight_layout()
    plt.savefig('fig_galectin_family_heatmap.png', dpi=150, bbox_inches='tight')
    plt.show()
    print(f"Figure saved: fig_galectin_family_heatmap.png  ({n_glycans} glycans, {n_lectins} lectins)")

    # ── Bonus: also print the numeric table so the values aren't lost ────
    print("\nBinding matrix (full numeric values):")
    print(galectin_measured.to_string())

else:
    print("No shared glycan columns found — heatmap skipped.")
```

## Section 9 — Search & Compare Utilities

Reusable functions for searching glycans across all lectins, and for motif-normalised cross-array comparison.

```python
# ── Section 9.1: search_glycan() utility ─────────────────────────────────
def search_glycan(query_pattern, top_n=10, min_score=0):
    """
    Search glycan_binding for all lectins that bind glycans matching query_pattern.
    Returns a DataFrame ranked by binding score.
    query_pattern : regex string in IUPAC-condensed notation
    """
    matching_cols = [
        c for c in glycan_cols
        if pd.Series([c]).str.contains(query_pattern, regex=True).any()
    ]
    if not matching_cols:
        print(f"No glycan columns match pattern: {query_pattern}")
        return pd.DataFrame()

    results = []
    for _, row in glycan_binding.iterrows():
        for col in matching_cols:
            score = row[col]
            if pd.notna(score) and score >= min_score:
                results.append({'protein': row['protein'], 'glycan': col, 'score': score})

    df = (pd.DataFrame(results)
            .sort_values('score', ascending=False)
            .reset_index(drop=True))
    return df.head(top_n)

# ── Example searches
print("=== All lectins binding LacNAc ===")
print(search_glycan(r'Gal\(b1-4\)GlcNAc', top_n=15))

print("\n=== All lectins binding T-antigen ===")
print(search_glycan(r'Gal\(b1-3\)GalNAc'))
```

```python
# ── Section 9.2: compare_lectins_by_motif() ──────────────────────────────
# Works even when lectins were profiled on non-overlapping CFG arrays.
def compare_lectins_by_motif(lectin_names, motif_pattern):
    """
    Compare multiple lectins on glycans sharing a structural motif.
    Uses motif presence (not direct column overlap) — works across array versions.
    """
    rows = []
    for lectin in lectin_names:
        row = glycan_binding[glycan_binding['protein'] == lectin]
        if row.empty:
            print(f"  Warning: '{lectin}' not found in glycan_binding")
            continue
        scores = pd.Series(row[glycan_cols].values[0], index=glycan_cols).dropna()
        motif_hits = scores[scores.index.str.contains(motif_pattern, regex=True)]
        for glycan, score in motif_hits.items():
            rows.append({'lectin': lectin, 'glycan': glycan, 'score': score})

    df = pd.DataFrame(rows)
    if df.empty:
        print("No overlapping data. These lectins may be on fully non-overlapping arrays.")
        print("This is expected for Gal3 vs Gal8c — use LectinOracle for unified comparison.")
        return df

    summary = df.groupby('lectin')['score'].agg(['mean', 'max', 'count'])
    summary.columns = ['mean_score', 'max_score', 'n_glycans_tested']
    print(f"Motif: {motif_pattern}")
    print(summary.to_string())
    return df

# β-galactoside comparison
compare_df = compare_lectins_by_motif(
    ['Galectin-3', 'Galectin-8 C-terminal CRD'],
    r'Gal\(b'
)
```
